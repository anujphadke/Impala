```python
import tensorflow as tf
```

```python
input_var=tf.Variable(initial_value=3.)
with tf.GradientTape() as tape:
  result = tf.square(input_var)

gradient = tape.gradient(result, input_var)
print(gradient)
```

watch the input variables on a tape. Easier to track variables for gradient computation that preemptively tracking everything. Trainable variables are tracked by default


```python
input_cost = tf.constant(4.)
with tf.GradientTape() as tape:
  tape.watch(input_cost)
  result = tf.sqrt(input_cost)
  gradient = tape.gradient(result, input_cost)
  print(gradient)
```

```python
time = tf.Variable(2.)
with tf.GradientTape() as outer_tape:
  with tf.GradientTape() as inner_tape:
    position =  4.9 * time ** 2
  speed = inner_tape.gradient(position, time)
acceleration = outer_tape.gradient(speed, time)

print(speed)
print(acceleration)
```
