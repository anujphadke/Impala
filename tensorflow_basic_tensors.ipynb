```python
import tensorflow as tf
x = tf.ones((2,2))
print(x.shape)
print(x.dtype)
print(x)
```

```python
x = tf.random.normal((5,1), mean=0, stddev=1)
print(x)
```

```python
y = tf.random.uniform((5,1), minval=0, maxval=5)
print(y)
```

```python
a=tf.Variable(initial_value=tf.random.normal((3,1)))
print(a)
```

```python
a.assign(tf.ones((3,1)))
print(a)
```

```python
m=tf.ones((3,1))
n=tf.square(m) + 5
o=tf.sqrt(n)
print(m)
print(n)
print(o)
```
